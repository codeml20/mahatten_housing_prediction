### Step 4: Clean and Convert Key Numerical Columns

```python
# 1. Clean and convert 'sale_price' to numeric
# First, convert to string to handle potential mixed types, then remove commas, and finally convert to numeric.
merged_df['sale_price'] = merged_df['sale_price'].astype(str).str.replace(',', '', regex=False)
merged_df['sale_price'] = pd.to_numeric(merged_df['sale_price'], errors='coerce')

# Handle columns that appear in both sales and pluto and prioritize sales data
# List of common numerical columns to consolidate
common_numeric_cols = [
    ('year_built', 'yearbuilt'),
    ('gross_square_feet', 'bldgarea'),
    ('land_square_feet', 'lotarea'),
    ('latitude', 'latitude'), # Assuming sales_latitude is more granular
    ('longitude', 'longitude') # Assuming sales_longitude is more granular
]

for sales_col, pluto_col in common_numeric_cols:
    # Define the column names in the merged_df
    sales_merged_col = f"{sales_col}"
    pluto_merged_col = f"{pluto_col}"

    if sales_merged_col in merged_df.columns and pluto_merged_col in merged_df.columns:
        # Convert both to numeric, coercing errors to NaN
        merged_df[sales_merged_col] = pd.to_numeric(merged_df[sales_merged_col].astype(str).str.replace(',', '', regex=False), errors='coerce')
        merged_df[pluto_merged_col] = pd.to_numeric(merged_df[pluto_merged_col].astype(str).str.replace(',', '', regex=False), errors='coerce')

        # Fill NaN values in the sales column with values from the pluto column
        merged_df[sales_merged_col] = merged_df[sales_merged_col].fillna(merged_df[pluto_merged_col])

        # Drop the pluto version of the column after consolidation
        merged_df = merged_df.drop(columns=[pluto_merged_col])
    elif pluto_merged_col in merged_df.columns and sales_merged_col not in merged_df.columns: # If only pluto version exists and is numerical
        merged_df[pluto_merged_col] = pd.to_numeric(merged_df[pluto_merged_col].astype(str).str.replace(',', '', regex=False), errors='coerce')
    elif sales_merged_col in merged_df.columns: # If only sales version exists and is numerical
        merged_df[sales_merged_col] = pd.to_numeric(merged_df[sales_merged_col].astype(str).str.replace(',', '', regex=False), errors='coerce')

# 3. Convert other key numerical columns from PLUTO
# The suffixes for pluto columns will still be there or no suffix if it was unique
pluto_numerical_cols = [
    'lotarea', 'bldgarea', 'comarea', 'resarea', 'officearea', 'retailarea',
    'garagearea', 'strgearea', 'factryarea', 'otherarea', 'numbldgs', 'numfloors',
    'unitsres', 'unitstotal', 'lotfront', 'lotdepth', 'bldgfront', 'bldgdepth',
    'assessland', 'assesstot', 'exemptland', 'exempttot', 'builtfar', 'residfar',
    'commfar', 'facilfar', 'xcoord', 'ycoord'
]

for col in pluto_numerical_cols:
    # Check if the column exists in the DataFrame
    if col in merged_df.columns:
        # Convert to string to handle various data formats, remove commas, then convert to numeric
        merged_df[col] = merged_df[col].astype(str).str.replace(',', '', regex=False)
        merged_df[col] = pd.to_numeric(merged_df[col], errors='coerce')

# For `unitsres`, `unitstotal` etc. from sales, convert to numeric as well
sales_units_cols = ['residential_units', 'commercial_units', 'total_units']
for col in sales_units_cols:
    if col in merged_df.columns:
        merged_df[col] = merged_df[col].astype(str).str.replace(',', '', regex=False)
        merged_df[col] = pd.to_numeric(merged_df[col], errors='coerce')

# Display info and head to verify conversions
print("\nDataFrame after numerical column cleaning and conversion:")
display(merged_df.head())
display(merged_df.info())
```

```python
import pandas as pd

# 1. Clean and convert 'sale_price' to numeric
# This line is redundant as sale_price was handled in the previous iteration of d3d62e05, but to be safe, keeping it here.
merged_df['sale_price'] = merged_df['sale_price'].astype(str).str.replace(',', '', regex=False)
merged_df['sale_price'] = pd.to_numeric(merged_df['sale_price'], errors='coerce')

# --- Start of explicit coordinate consolidation ---
# Ensure 'latitude' column is present and numeric, prioritizing sales data
if 'latitude_sales' in merged_df.columns:
    merged_df['latitude'] = pd.to_numeric(merged_df['latitude_sales'].astype(str).str.replace(',', '', regex=False), errors='coerce')
    merged_df = merged_df.drop(columns=['latitude_sales'])
elif 'latitude' in merged_df.columns: # If not suffixed, means it was already just 'latitude' in sales or previous step
    merged_df['latitude'] = pd.to_numeric(merged_df['latitude'].astype(str).str.replace(',', '', regex=False), errors='coerce')
else:
    merged_df['latitude'] = pd.NA # Fallback if no sales latitude

if 'latitude_pluto' in merged_df.columns:
    pluto_lat_col = pd.to_numeric(merged_df['latitude_pluto'].astype(str).str.replace(',', '', regex=False), errors='coerce')
    merged_df['latitude'] = merged_df['latitude'].fillna(pluto_lat_col)
    merged_df = merged_df.drop(columns=['latitude_pluto'])

# Ensure 'longitude' column is present and numeric, prioritizing sales data
if 'longitude_sales' in merged_df.columns:
    merged_df['longitude'] = pd.to_numeric(merged_df['longitude_sales'].astype(str).str.replace(',', '', regex=False), errors='coerce')
    merged_df = merged_df.drop(columns=['longitude_sales'])
elif 'longitude' in merged_df.columns: # If not suffixed
    merged_df['longitude'] = pd.to_numeric(merged_df['longitude'].astype(str).str.replace(',', '', regex=False), errors='coerce')
else:
    merged_df['longitude'] = pd.NA

if 'longitude_pluto' in merged_df.columns:
    pluto_lon_col = pd.to_numeric(merged_df['longitude_pluto'].astype(str).str.replace(',', '', regex=False), errors='coerce')
    merged_df['longitude'] = merged_df['longitude'].fillna(pluto_lon_col)
    merged_df = merged_df.drop(columns=['longitude_pluto'])
# --- End of explicit coordinate consolidation ---


# Now, handle other common numerical columns. These were not suffixed in the original merge.
consolidate_map_other_cols = {
    'year_built': ('year_built', 'yearbuilt'), # sales_manhattan had 'year_built', pluto had 'yearbuilt'
    'gross_square_feet': ('gross_square_feet', 'bldgarea'), # sales_manhattan had 'gross_square_feet', pluto had 'bldgarea'
    'land_square_feet': ('land_square_feet', 'lotarea'), # sales_manhattan had 'land_square_feet', pluto had 'lotarea'
}

for final_col, (primary_col, secondary_col) in consolidate_map_other_cols.items():
    # Attempt to use the 'sales' version first
    if primary_col in merged_df.columns:
        merged_df[final_col] = pd.to_numeric(merged_df[primary_col].astype(str).str.replace(',', '', regex=False), errors='coerce')
        if primary_col != final_col: # Drop original if it's different from the final name
             merged_df = merged_df.drop(columns=[primary_col])
    else:
        merged_df[final_col] = pd.NA

    # Fill NaNs with the 'pluto' version if available
    if secondary_col in merged_df.columns:
        secondary_numeric = pd.to_numeric(merged_df[secondary_col].astype(str).str.replace(',', '', regex=False), errors='coerce')
        merged_df[final_col] = merged_df[final_col].fillna(secondary_numeric)
        if secondary_col != final_col:
             merged_df = merged_df.drop(columns=[secondary_col]) # Drop the secondary column


# Identify other numerical columns from PLUTO and sales that were not part of the consolidation above.
# These typically keep their original names if there was no collision or are unique to one dataset.
pluto_numerical_cols = [
    'comarea', 'resarea', 'officearea', 'retailarea',
    'garagearea', 'strgearea', 'factryarea', 'otherarea', 'numbldgs', 'numfloors',
    'unitsres', 'unitstotal', 'lotfront', 'lotdepth', 'bldgfront', 'bldgdepth',
    'assessland', 'assesstot', 'exemptland', 'exempttot', 'builtfar', 'residfar',
    'commfar', 'facilfar', 'xcoord', 'ycoord'
]

sales_units_cols = ['residential_units', 'commercial_units', 'total_units']

# Combine all numerical columns to ensure they are processed.
all_remaining_numerical_cols = list(set(pluto_numerical_cols + sales_units_cols))

for col in all_remaining_numerical_cols:
    # Ensure that we only process columns that exist in the DataFrame
    # and were NOT already handled by the `consolidate_map_other_cols` or the coordinate consolidation.
    if col in merged_df.columns and col not in consolidate_map_other_cols.keys() and col not in ['latitude', 'longitude']:
        merged_df[col] = merged_df[col].astype(str).str.replace(',', '', regex=False)
        merged_df[col] = pd.to_numeric(merged_df[col], errors='coerce')

# Display info and head to verify conversions
print("\nDataFrame after numerical column cleaning and conversion:")
display(merged_df.head())
display(merged_df.info())
```

```python
from math import radians, sin, cos, sqrt, atan2

def haversine_distance(lat1, lon1, lat2, lon2):
    R = 6371  # Radius of Earth in kilometers

    lat1, lon1, lat2, lon2 = map(radians, [lat1, lon1, lat2, lon2])

    dlon = lon2 - lon1
    dlat = lat2 - lat1

    a = sin(dlat / 2)**2 + cos(lat1) * cos(lat2) * sin(dlon / 2)**2
    c = 2 * atan2(sqrt(a), sqrt(1 - a))

    distance = R * c
    return distance

# Convert MTA coordinates to numeric
mta['gtfs_latitude'] = pd.to_numeric(mta['gtfs_latitude'], errors='coerce')
mta['gtfs_longitude'] = pd.to_numeric(mta['gtfs_longitude'], errors='coerce')

# Drop rows with NaN coordinates in mta, if any
mta_clean = mta.dropna(subset=['gtfs_latitude', 'gtfs_longitude']).copy()

# The 'latitude' and 'longitude' columns in merged_df are now handled by the previous cell 'd3d6e205'
# We only need to ensure they exist and drop NaNs for distance calculation.

# Drop rows with NaN coordinates in merged_df for distance calculation
merged_df_geo = merged_df.dropna(subset=['latitude', 'longitude']).copy()

# Calculate distance to nearest subway station for each property
def get_min_subway_distance(property_lat, property_lon, mta_stations_df):
    if pd.isna(property_lat) or pd.isna(property_lon):
        return None

    distances = mta_stations_df.apply(
        lambda row: haversine_distance(property_lat, property_lon, row['gtfs_latitude'], row['gtfs_longitude']),
        axis=1
    )
    return distances.min()

print("Calculating distances to nearest subway stations. This might take a moment...")

# Placeholder for the new column (will be overwritten)
merged_df['distance_to_subway_km'] = None # Initialize with None

# Apply for the entire dataset
merged_df['distance_to_subway_km'] = merged_df_geo.apply(
    lambda row: get_min_subway_distance(row['latitude'], row['longitude'], mta_clean),
    axis=1
)

print("Distance calculation complete. Displaying head and info.")

# Display head and info to show the new column
display(merged_df.head())
display(merged_df.info())
```

```python
import numpy as np
from scipy.spatial import KDTree

# 1. Define Express Lines (Common NYC Express Routes)
# Note: This is a heuristic; in NYC, lines like 2,3,4,5,A,B,D,Q often have express segments.
express_lines_query = "daytime_routes.str.contains('2') | daytime_routes.str.contains('3') | \
                       daytime_routes.str.contains('4') | daytime_routes.str.contains('5') | \
                       daytime_routes.str.contains('A') | daytime_routes.str.contains('D') | \
                       daytime_routes.str.contains('B') | daytime_routes.str.contains('Q')"

mta_clean['is_express'] = mta_clean.eval(express_lines_query)
express_stations = mta_clean[mta_clean['is_express']].copy()

# 2. Conversion factor: 1 degree lat is ~111km.
# For small distances like NYC, we can approximate using a Euclidean-based KDTree
# on coordinates and then scale to kilometers.
# Scaling factors for NYC latitude (approx 40.7)
LAT_TO_KM = 111.0
LON_TO_KM = 111.0 * np.cos(np.radians(40.73))

def get_scaled_coords(df, lat_col, lon_col):
    # Scale degrees to kilometers for roughly accurate Euclidean distance in a local area
    coords = df[[lat_col, lon_col]].values.astype(float)
    coords[:, 0] *= LAT_TO_KM
    coords[:, 1] *= LON_TO_KM
    return coords

# Prepare coordinates
prop_coords_km = get_scaled_coords(merged_df_geo, 'latitude', 'longitude')
all_st_coords_km = get_scaled_coords(mta_clean, 'gtfs_latitude', 'gtfs_longitude')
exp_st_coords_km = get_scaled_coords(express_stations, 'gtfs_latitude', 'gtfs_longitude')

# Build Trees
tree_all = KDTree(all_st_coords_km)
tree_exp = KDTree(exp_st_coords_km)

# 3. Calculations
# Nearest Station Distances
dist_all, idx_all = tree_all.query(prop_coords_km)
dist_exp, _ = tree_exp.query(prop_coords_km)

# Counts within Radii
# stations within 500m (0.5 km)
count_500m = [len(i) for i in tree_all.query_ball_point(prop_coords_km, r=0.5)]

# Subway lines within 1km
# We need to look up which stations are in range, then get unique lines
indices_1km = tree_all.query_ball_point(prop_coords_km, r=1.0)
unique_lines_1km = []
for neighbor_indices in indices_1km:
    if neighbor_indices:
        routes = mta_clean.iloc[neighbor_indices]['daytime_routes'].str.split().explode().unique()
        unique_lines_1km.append(len(routes))
    else:
        unique_lines_1km.append(0)

# 4. Assign back to the DataFrame
# Use the index of merged_df_geo to map back to merged_df correctly
geo_idx = merged_df_geo.index
merged_df.loc[geo_idx, 'dist_nearest_st_km'] = dist_all
merged_df.loc[geo_idx, 'num_st_500m'] = count_500m
merged_df.loc[geo_idx, 'num_lines_1km'] = unique_lines_1km
merged_df.loc[geo_idx, 'is_walking_dist'] = (dist_all <= 0.8).astype(int)
merged_df.loc[geo_idx, 'dist_express_st_km'] = dist_exp

print("Optimized geospatial features calculated successfully.")
display(merged_df[['address_sales', 'dist_nearest_st_km', 'num_st_500m', 'num_lines_1km', 'is_walking_dist', 'dist_express_st_km']].head())
```

```python
from sklearn.model_selection import KFold
from sklearn.inspection import permutation_importance

def smooth_target_encode(train_df, test_df, column, target_values, k=5, m=20):
    # target_values is the y_tr series
    train_df[f'{column}_reg'] = 0.0
    test_df[f'{column}_reg'] = 0.0

    kf = KFold(n_splits=k, shuffle=True, random_state=42)
    global_mean = target_values.mean()

    # K-Fold out-of-fold encoding for Training
    for train_idx, val_idx in kf.split(train_df):
        # Use current target values for this specific split
        y_t = target_values.iloc[train_idx]
        X_t = train_df.iloc[train_idx]

        # Create temporary agg table
        temp_df = pd.DataFrame({'col': X_t[column], 'target': y_t})
        agg = temp_df.groupby('col')['target'].agg(['count', 'mean'])

        smooth = (agg['mean'] * agg['count'] + global_mean * m) / (agg['count'] + m)
        train_df.loc[train_df.index[val_idx], f'{column}_reg'] = train_df.loc[train_df.index[val_idx], column].map(smooth)

    # Encode Test data using full Train stats
    full_temp = pd.DataFrame({'col': train_df[column], 'target': target_values})
    full_agg = full_temp.groupby('col')['target'].agg(['count', 'mean'])
    full_smooth = (full_agg['mean'] * full_agg['count'] + global_mean * m) / (full_agg['count'] + m)

    test_df[f'{column}_reg'] = test_df[column].map(full_smooth)

    # Fill NaNs
    train_df[f'{column}_reg'] = train_df[f'{column}_reg'].fillna(global_mean)
    test_df[f'{column}_reg'] = test_df[f'{column}_reg'].fillna(global_mean)
    return train_df, test_df

# Prepare fresh split data
X_tr_reg = X_tr.copy()
X_te_reg = X_te.copy()

# Re-attach raw strings for encoding
for col in cat_to_encode:
    X_tr_reg[col] = merged_df_cleaned.loc[X_tr_reg.index, col].astype(str)
    X_te_reg[col] = merged_df_cleaned.loc[X_te_reg.index, col].astype(str)

# Apply Regularized Smoothing using y_tr as the target reference
for col in cat_to_encode:
    X_tr_reg, X_te_reg = smooth_target_encode(X_tr_reg, X_te_reg, col, y_tr, m=20)
    X_tr_reg = X_tr_reg.drop(columns=[col])
    X_te_reg = X_te_reg.drop(columns=[col])

# Drop old LOO columns
X_tr_reg = X_tr_reg.drop(columns=[c for c in X_tr_reg.columns if '_loo' in c], errors='ignore')
X_te_reg = X_te_reg.drop(columns=[c for c in X_te_reg.columns if '_loo' in c], errors='ignore')

# Train and evaluate
xgb_reg = XGBRegressor(n_estimators=200, learning_rate=0.05, max_depth=5, random_state=42)
xgb_reg.fit(X_tr_reg, y_tr)

result_reg = permutation_importance(xgb_reg, X_te_reg, y_te, n_repeats=5, random_state=42)
perm_df_reg = pd.DataFrame({'feature': X_te_reg.columns, 'importance': result_reg.importances_mean}).sort_values(by='importance', ascending=False)

plt.figure(figsize=(10, 8))
sns.barplot(x='importance', y='feature', data=perm_df_reg.head(15), palette='viridis')
plt.title('Regularized Importance: Categories Smoothed Toward Global Mean')
plt.show()

print(f"Regularized Model R-squared: {xgb_reg.score(X_te_reg, y_te):.4f}")
```

```python
import numpy as np
from scipy.spatial import KDTree

# 1. Define the radius for our 'Hub' (500 meters)
HUB_RADIUS_KM = 0.5

# 2. Extract coordinates (using correct names identified in kernel state)
prop_coords = merged_df_ml[['latitude_sales', 'longitude_sales']].values
station_coords = mta_clean[['gtfs_latitude', 'gtfs_longitude']].values

# 3. Scaling for Euclidean distance in NYC area
LAT_TO_KM = 111.0
LON_TO_KM = 111.0 * np.cos(np.radians(40.73))

def scale_coords(coords):
    c = coords.copy().astype(float)
    c[:, 0] *= LAT_TO_KM
    c[:, 1] *= LON_TO_KM
    return c

tree = KDTree(scale_coords(station_coords))
prop_coords_scaled = scale_coords(prop_coords)

# 4. Find all station indices within 500m
indices_within_hub = tree.query_ball_point(prop_coords_scaled, r=HUB_RADIUS_KM)

# 5. Calculate unique routes
def count_unique_routes(indices, mta_df):
    if not indices:
        return 0
    routes_str = ' '.join(mta_df.iloc[indices]['daytime_routes'].values)
    return len(set(routes_str.split()))

unique_lines_count = [count_unique_routes(idx, mta_clean) for idx in indices_within_hub]

# 6. Create the features
merged_df_ml['num_unique_lines_500m'] = unique_lines_count
merged_df_ml['transit_hub_score'] = merged_df_ml['num_unique_lines_500m'] / (merged_df_ml['dist_nearest_st_km'] + 0.1)

print(f"Transit Hub Score calculated. Range: {merged_df_ml['transit_hub_score'].min():.2f} to {merged_df_ml['transit_hub_score'].max():.2f}")
display(merged_df_ml[['num_unique_lines_500m', 'transit_hub_score', 'dist_nearest_st_km']].head())
```

```python
import pandas as pd
import numpy as np
import re
import shap
from xgboost import XGBRegressor
from sklearn.feature_selection import RFECV
from sklearn.model_selection import KFold

# 1. Algorithmic Cleanup: Remove any column containing technical 'code', 'id', 'bbl', or 'prox'
# This targets things like bsmtcode, proxcode, bctcb2020 automatically.
pattern = re.compile(r'code|id$|bbl|prox|coord|tract|block|lot|suffix|census_tract|nta|address', re.IGNORECASE)
technical_cols = [c for c in X_practical.columns if pattern.search(c)]
X_algo = X_practical.drop(columns=technical_cols)

# Ensure Census features are preserved explicitly as they are our 'practical' targets
census_cols = ['median_household_income', 'median_home_value', 'total_population', 'median_gross_rent']
for col in census_cols:
    if col in ml_df.columns and col not in X_algo.columns:
        X_algo[col] = ml_df[col]

# 2. Recursive Feature Elimination (RFECV)
# Let the model decide the optimal number of features based on cross-validation
xgb_selector = XGBRegressor(n_estimators=50, max_depth=5, learning_rate=0.1, random_state=42)
selector = RFECV(estimator=xgb_selector, step=5, cv=KFold(3), scoring='r2', n_jobs=-1)
selector.fit(X_algo, y_practical)

# Select the features
selected_features = X_algo.columns[selector.support_].tolist()
X_final = X_algo[selected_features]

# 3. Final Model Training
xgb_final = XGBRegressor(n_estimators=200, learning_rate=0.05, max_depth=6, random_state=42)
xgb_final.fit(X_final, y_practical)

print(f"Selected {len(selected_features)} practical features.")
print(f"Final R2: {xgb_final.score(X_final, y_practical):.4f}")
```

```python
# 4. SHAP Explanation: Answering why Census features rank where they do
explainer = shap.TreeExplainer(xgb_final)
shap_values = explainer.shap_values(X_final)

# Plot SHAP summary to show the 'Impact' (Practical Importance)
plt.title("SHAP Value Analysis: Practical Impact of Features")
shap.summary_plot(shap_values, X_final, max_display=20, plot_type="bar")
```

```python
# 5. Direct Census Contribution Breakdown
# We check how much 'Median Household Income' and 'Median Home Value' specifically shift predictions
shap_importance = np.abs(shap_values).mean(axis=0)
shap_df = pd.DataFrame({'feature': X_final.columns, 'shap_importance': shap_importance}).sort_values(by='shap_importance', ascending=False)

print("Detailed Census Impact Comparison:")
display(shap_df[shap_df['feature'].isin(census_cols)])

print("\nTop 10 most impactful features after Algorithmic selection:")
display(shap_df.head(10))
```

### Causal Inference: Estimating the 'Neighborhood Premium'
We define a Causal Model where:
*   **Treatment**: `high_income_neighborhood` (Binary: Above/Below Median)
*   **Outcome**: `sale_price` (Log-transformed)
*   **Common Causes (Confounders)**: `gross_square_feet`, `year_built`, `total_units`, `tax_class_as_of_final_roll`

```python
import dowhy
from dowhy import CausalModel
import pandas as pd
import numpy as np

# Prepare data for causal analysis
causal_df = ml_df.copy()
income_threshold = causal_df['median_household_income'].median()
causal_df['is_high_income'] = (causal_df['median_household_income'] > income_threshold).astype(int)
causal_df['log_price'] = np.log1p(causal_df['sale_price'])

# Define the causal graph
model = CausalModel(
    data=causal_df,
    treatment='is_high_income',
    outcome='log_price',
    common_causes=['gross_square_feet', 'year_built', 'total_units', 'tax_class_as_of_final_roll']
)

# 1. Identify the effect
identified_estimand = model.identify_effect(proceed_when_unidentifiable=True)

# 2. Estimate the effect using Propensity Score Stratification
estimate = model.estimate_effect(identified_estimand,
                                 method_name="backdoor.propensity_score_stratification")

print(f"Causal Estimate (ATE): {estimate.value:.4f}")

# 3. Refutation Tests: Validating the causal estimate
print("\nRunning Refutation Tests...")

# Random Common Cause: Does adding a random confounder change the estimate? (Robust if no change)
refute_random = model.refute_estimate(identified_estimand, estimate, method_name="random_common_cause")
print(f"Random Common Cause Effect: {refute_random.estimated_effect:.4f}")

# Placebo Treatment: If we randomize the treatment, does the effect go to zero? (Robust if near 0)
refute_placebo = model.refute_estimate(identified_estimand, estimate, method_name="placebo_treatment_refuter", placebo_type="permute")
print(f"Placebo Treatment Effect: {refute_placebo.estimated_effect:.4f}")
```

### Causal Refinement: Linear Backdoor Estimation
To satisfy research-grade rigor, we move from Propensity Stratification (which can be unstable with many confounders) to a **Linear Regression Estimator**. We will also run a 'Random Common Cause' and 'Placebo' test specifically to check if the effect vanishes when the treatment is randomized—a critical check for causal validity.

```python
from dowhy import CausalModel
import numpy as np

# Define Model using Linear Regression approach
causal_model = CausalModel(
    data=causal_df,
    treatment='is_high_income',
    outcome='log_price',
    common_causes=['gross_square_feet', 'year_built', 'total_units', 'tax_class_as_of_final_roll']
)

# Identify and Estimate
identified_estimand = causal_model.identify_effect(proceed_when_unidentifiable=True)
linear_estimate = causal_model.estimate_effect(
    identified_estimand,
    method_name="backdoor.linear_regression",
    test_significance=True
)

print(f"Refined Causal Estimate (Linear ATE): {linear_estimate.value:.4f}")

print("\n--- Causal Validation Tests ---")

# Placebo Test: Effect should be ~0 if treatment is permuted
placebo_refutation = causal_model.refute_estimate(
    identified_estimand, linear_estimate,
    method_name="placebo_treatment_refuter",
    placebo_type="permute",
    num_simulations=10
)
print(f"Placebo Effect (Target: ~0): {placebo_refutation.estimated_effect:.4f}")

# Random Common Cause: Effect should remain stable
random_cause_refutation = causal_model.refute_estimate(
    identified_estimand, linear_estimate,
    method_name="random_common_cause"
)
print(f"Random Common Cause Effect (Target: Stable): {random_cause_refutation.estimated_effect:.4f}")
```
